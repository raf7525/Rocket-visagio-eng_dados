```python
from pyspark.sql import functions as F
from pyspark.sql.window import Window

# 1. Dedup de reingestão
window_dedup_ingestao = Window.partitionBy("id").orderBy(F.col("ingestion_datetime").desc())

df_base = (
    spark.table("catalogo_filmes.bronze.tb_movies_info")
    .withColumn("_rn", F.row_number().over(window_dedup_ingestao))
    .filter(F.col("_rn") == 1)
    .drop("_rn")
)

# 2. Chaves normalizadas
df_base = (
    df_base
    .withColumn("titulo_norm", F.trim(F.regexp_replace(F.lower(F.col("title")), r"[^a-z0-9à-ÿ]+", " ")))
    .withColumn("sinopse_norm", F.trim(F.regexp_replace(F.lower(F.col("overview")), r"[^a-z0-9à-ÿ]+", " ")))
    .withColumn(
        "data_norm",
        F.coalesce(
            F.expr("try_to_date(release_date, 'yyyy-MM-dd')"),
            F.expr("try_to_date(release_date, 'yyyy/MM/dd')"),
            F.expr("try_to_date(release_date, 'dd/MM/yyyy')"),
            F.expr("try_to_date(release_date, 'dd-MM-yyyy')"),
            F.expr("try_to_date(release_date, 'MM/dd/yyyy')"),
            F.expr("try_to_date(release_date, 'MM-dd-yyyy')"),
        )
    )
    .withColumn("ano_norm", F.year(F.col("data_norm")))
    # assinatura numérica do título: "chapter 2" -> "2", "die hart die harter" -> ""
    .withColumn("digitos_titulo", F.regexp_replace(F.col("titulo_norm"), r"[^0-9]", ""))
    # prefixo: as 2 primeiras palavras do título normalizado
    .withColumn("prefixo_titulo", F.array_join(F.slice(F.split(F.col("titulo_norm"), " "), 1, 2), " "))
    .withColumn("chave_sinopse", F.when(F.length(F.col("sinopse_norm")) >= 80, F.col("sinopse_norm")))
)

# PASSADA 1 — título exato + data exata (separa homônimos como os vários "Silence")
df_p1 = df_base.withColumn(
    "canon_1",
    F.when(
        (F.col("titulo_norm").isNotNull()) & (F.col("titulo_norm") != "") & (F.col("data_norm").isNotNull()),
        F.min("id").over(Window.partitionBy("titulo_norm", "data_norm"))
    ).otherwise(F.col("id"))
)

# PASSADA 2a — mesma sinopse + mesma assinatura numérica
df_p2a = df_p1.withColumn(
    "canon_2a",
    F.when(
        F.col("chave_sinopse").isNotNull(),
        F.min("canon_1").over(Window.partitionBy("chave_sinopse", "digitos_titulo"))
    ).otherwise(F.col("canon_1"))
)

# PASSADA 2b — título SEM dígito adota o grupo da mesma sinopse.
# Assim "die hart die harter" se junta a "die hart 2 die harter", mas
# "part 1" e "part 2" (ambos COM dígito) continuam intocados.
df_p2b = df_p2a.withColumn(
    "canon_2",
    F.when(
        F.col("chave_sinopse").isNotNull() & (F.col("digitos_titulo") == ""),
        F.min("canon_2a").over(Window.partitionBy("chave_sinopse"))
    ).otherwise(F.col("canon_2a"))
)

# PASSADA 3 — registros SEM sinopse utilizável adotam o grupo de quem compartilha
# prefixo de título + ano + assinatura numérica. Quem tem sinopse própria não é tocado,
# então homônimos com sinopses distintas seguem separados.
df_p3 = (
    df_p2b
    .withColumn(
        "_min_grupo",
        F.min("canon_2").over(Window.partitionBy("prefixo_titulo", "ano_norm", "digitos_titulo"))
    )
    .withColumn(
        "id_filme_canonico",
        F.when(
            F.col("chave_sinopse").isNull()
            & (F.col("prefixo_titulo") != "")
            & F.col("ano_norm").isNotNull(),
            F.col("_min_grupo")
        ).otherwise(F.col("canon_2"))
    )
)

# PASSADA 4 — propagação por transitividade: todos os registros que a passada 1
# considerou o mesmo filme (título e data idênticos) devem terminar no mesmo grupo,
# mesmo que só parte deles tenha migrado nas passadas 2 e 3.
df_p4 = df_p3.withColumn(
    "id_filme_canonico",
    F.min("id_filme_canonico").over(Window.partitionBy("canon_1"))
)

df_mapa_id_filme = (
    df_p4
    .select(F.col("id").alias("id_filme_original"), "id_filme_canonico")
    .distinct()
)

df_mapa_id_filme.write.format("delta").mode("overwrite").option("overwriteSchema", "true").saveAsTable("catalogo_filmes.silver.tb_mapa_id_filme")

display(df_mapa_id_filme.filter(F.col("id_filme_original") != F.col("id_filme_canonico")))
```

```python
from pyspark.sql import functions as F
from pyspark.sql.window import Window

df_mapa = spark.table("catalogo_filmes.silver.tb_mapa_id_filme")


def limpar_texto(coluna):
    """Remove resíduo de escape do CSV quebrado (barras e aspas repetidas) e normaliza espaços."""
    sem_barra = F.regexp_replace(coluna, r"\\", "")
    aspas_colapsadas = F.regexp_replace(sem_barra, r'"{2,}', '"')
    sem_borda = F.regexp_replace(aspas_colapsadas, r'^["\s]+|["\s]+$', "")
    return F.trim(F.regexp_replace(sem_borda, r"\s+", " "))


# 1. Dedup de REINGESTÃO
window_dedup_ingestao = Window.partitionBy("id").orderBy(F.col("ingestion_datetime").desc())

df_bronze_raw = (
    spark.table("catalogo_filmes.bronze.tb_movies_info")
    .withColumn("_rn", F.row_number().over(window_dedup_ingestao))
    .filter(F.col("_rn") == 1)
    .drop("_rn")
)

# 2. Traduz o id original para o id CANÔNICO
df_bronze_movies = (
    df_bronze_raw
    .join(df_mapa, df_bronze_raw.id == df_mapa.id_filme_original, "left")
    .withColumn("id", F.coalesce(F.col("id_filme_canonico"), df_bronze_raw.id))
    .drop("id_filme_canonico", "id_filme_original")
)

# 3. Segundo dedup, agora pelo id canônico
window_dedup_canonico = Window.partitionBy("id").orderBy(F.col("ingestion_datetime").desc())

df_silver_movies = (
    df_bronze_movies

    .withColumn("_rn2", F.row_number().over(window_dedup_canonico))
    .filter(F.col("_rn2") == 1)
    .drop("_rn2")

    .withColumn("title", F.trim(F.col("title")))
    .withColumn("original_title", F.trim(F.col("original_title")))
    .withColumn(
        "_status_clean",
        F.lower(F.trim(F.regexp_replace(F.regexp_replace(F.col("status"), r"[-_]+", " "), r"\s+", " ")))
    )
    .withColumn(
        "status",
        F.when(F.col("_status_clean") == "released", "Lançado")
         .when(F.col("_status_clean") == "post production", "Pós-Produção")
         .when(F.col("_status_clean") == "in production", "Em Produção")
         .when(F.col("_status_clean") == "planned", "Planejado")
         .when(F.col("_status_clean") == "rumored", "Rumores")
         .when(F.col("_status_clean") == "canceled", "Cancelado")
         .otherwise("Não Informado")
    )
    .drop("_status_clean")

    .select(
        "id", "title", "original_title", "release_date",
        "runtime", "original_language", "status", "overview", "tagline"
    )

    .withColumnRenamed("id", "id_filme")
    .withColumnRenamed("title", "titulo")
    .withColumnRenamed("original_title", "titulo_original")
    .withColumnRenamed("release_date", "data_lancamento")
    .withColumnRenamed("runtime", "duracao_minutos")
    .withColumnRenamed("original_language", "idioma_original")
    .withColumnRenamed("status", "status_filme")
    .withColumnRenamed("overview", "sinopse")
    .withColumnRenamed("tagline", "frase_divulgacao")

    # ---- DETECÇÃO DE LINHA CORROMPIDA (feita ANTES de limpar os campos) ----
    # Sinal 1: idioma fora do padrão ISO 639-1 (2 letras) indica desalinhamento de coluna
    .withColumn(
        "_idioma_valido",
        F.lower(F.trim(F.col("idioma_original"))).rlike(r"^[a-z]{2}$")
    )
    # Sinal 2: a sinopse contém uma "linha de CSV grudada" — assinatura típica é
    # data ISO + duração + status colados por vírgula (ex: "...tr2019-09-06,115,released,")
    .withColumn(
        "_sinopse_suspeita",
        F.col("sinopse").rlike(r"\d{4}-\d{2}-\d{2},\d+,") |
        F.col("sinopse").rlike(r"(?i),\s*(released|post production|in production|planned|rumored|canceled)\s*,")
    )

    # ---- LIMPEZA DE ESCAPE nos campos de texto ----
    .withColumn("titulo", limpar_texto(F.col("titulo")))
    .withColumn("titulo_original", limpar_texto(F.col("titulo_original")))
    .withColumn("sinopse", limpar_texto(F.col("sinopse")))
    .withColumn("frase_divulgacao", limpar_texto(F.col("frase_divulgacao")))

    # ---- IDIOMA: só sobrevive código ISO de 2 letras ----
    .withColumn(
        "idioma_original",
        F.when(F.col("_idioma_valido"), F.lower(F.trim(F.col("idioma_original")))).otherwise(None)
    )

    # ---- SINOPSE: invalida quando a linha está corrompida ou o texto ficou vazio ----
    .withColumn(
        "sinopse",
        F.when(
            (~F.col("_idioma_valido")) | F.col("_sinopse_suspeita") | (F.col("sinopse") == ""),
            None
        ).otherwise(F.col("sinopse"))
    )
    .drop("_idioma_valido", "_sinopse_suspeita")

    .withColumn(
        "data_lancamento",
        F.coalesce(
            F.expr("try_to_date(data_lancamento, 'yyyy-MM-dd')"),
            F.expr("try_to_date(data_lancamento, 'yyyy/MM/dd')"),
            F.expr("try_to_date(data_lancamento, 'dd/MM/yyyy')"),
            F.expr("try_to_date(data_lancamento, 'dd-MM-yyyy')"),
            F.expr("try_to_date(data_lancamento, 'MM/dd/yyyy')"),
            F.expr("try_to_date(data_lancamento, 'MM-dd-yyyy')"),
        )
    )

    .withColumn("duracao_minutos", F.expr("try_cast(duracao_minutos as int)"))

    # duração 0 ou negativa é placeholder de ausência, não filme de zero minutos
    .withColumn(
        "duracao_minutos",
        F.when(F.col("duracao_minutos") <= 0, None).otherwise(F.col("duracao_minutos"))
    )

    .withColumn("ano_lancamento", F.year(F.col("data_lancamento")))
)

df_silver_movies.write.format("delta").mode("overwrite").option("overwriteSchema", "true").saveAsTable("catalogo_filmes.silver.tb_info_filmes")

display(df_silver_movies)
```

```python
df_bronze_cotacao = spark.table("catalogo_filmes.bronze.tb_cotacao_dolar")

# 1. A API retorna timestamp completo (ex: "2026-09-14 13:10:08.144425"), mas queremos
#    1 valor por DIA. Convertemos pra data pura e, se houver mais de um registro no
#    mesmo dia, ficamos com o último (F.last dentro do groupBy).
df_cotacao_diaria = (
    df_bronze_cotacao
    .withColumn("data", F.to_date(F.col("dataHoraCotacao")))
    .groupBy("data")
    .agg(F.last("cotacaoCompra").alias("cotacao_compra"))
)

# 2. Descobre o período coberto (menor e maior data que você tem cotação real)
data_min, data_max = df_cotacao_diaria.selectExpr("min(data)", "max(data)").first()

# 3. Gera um "calendário" com TODOS os dias entre data_min e data_max, incluindo
#    finais de semana/feriados que a API não retorna. sequence() cria um array de
#    datas consecutivas, e explode() transforma cada item do array numa linha.
df_calendario = spark.sql(
    f"SELECT explode(sequence(to_date('{data_min}'), to_date('{data_max}'), interval 1 day)) AS data"
)

# 4. Junta o calendário completo com as cotações reais. Dias sem cotação
#    (fins de semana/feriados) ficam com cotacao_compra = NULL depois desse join.
df_calendario_cotacao = df_calendario.join(df_cotacao_diaria, on="data", how="left")

# 5. FORWARD FILL: para cada linha, olhando as datas em ordem crescente,
#    pega o último valor NÃO NULO visto "até agora" (rowsBetween(unboundedPreceding, 0)
#    significa "de todas as linhas anteriores até a linha atual").
#    F.last(..., ignorenulls=True) ignora os NULLs no caminho e repete o último valor real.
window_ffill = Window.orderBy("data").rowsBetween(Window.unboundedPreceding, 0)

df_silver_cotacao = df_calendario_cotacao.withColumn(
    "cotacao_compra",
    F.last("cotacao_compra", ignorenulls=True).over(window_ffill)
)

df_silver_cotacao.write.format("delta").mode("overwrite").saveAsTable("catalogo_filmes.silver.tb_cotacao_dolar")

display(df_silver_cotacao.orderBy("data"))
```

```python
def limpar_valor_monetario(df, coluna_origem, coluna_destino):
    """Converte valor monetário sujo em decimal, tratando notação abreviada (K/M/B),
    prefixo de moeda (USD, $, R$) e textos de ausência."""

    valores_ausentes = ["unknown", "não informado", "nao informado", "n/a", "na", ""]

    # Formato aceito: prefixo de moeda opcional + dígitos/separadores + sufixo K/M/B opcional.
    # Qualquer coisa fora disso é texto vazado por Column Shift.
    padrao_valido = r"^(USD|US\$|\$|R\$|BRL)?[\s0-9.,]+[KMB]?$"

    return (
        df
        # 1. textos conhecidos de ausência -> NULL
        .withColumn(
            f"_{coluna_destino}_t1",
            F.when(F.lower(F.trim(F.col(coluna_origem))).isin(valores_ausentes), None)
             .otherwise(F.upper(F.trim(F.col(coluna_origem))))
        )
        # 2. só sobrevive o que tem CARA de valor monetário
        .withColumn(
            f"_{coluna_destino}_t2",
            F.when(F.col(f"_{coluna_destino}_t1").rlike(padrao_valido), F.col(f"_{coluna_destino}_t1"))
        )
        # 3. multiplicador pelo sufixo: 34.0M = 34 milhões, 800K = 800 mil, 1.2B = 1,2 bilhão.
        #    Exige dígito antes da letra, pra não confundir com o "B" de "BRL".
        .withColumn(
            f"_{coluna_destino}_mult",
            F.when(F.col(f"_{coluna_destino}_t2").rlike(r"\d\s*B$"), F.lit(1000000000))
             .when(F.col(f"_{coluna_destino}_t2").rlike(r"\d\s*M$"), F.lit(1000000))
             .when(F.col(f"_{coluna_destino}_t2").rlike(r"\d\s*K$"), F.lit(1000))
             .otherwise(F.lit(1))
        )
        # 4. remove vírgula de milhar e tudo que não for dígito/ponto/sinal
        .withColumn(
            f"_{coluna_destino}_t3",
            F.regexp_replace(F.regexp_replace(F.col(f"_{coluna_destino}_t2"), ",", ""), r"[^0-9.\-]", "")
        )
        # 5. converte e aplica o multiplicador
        .withColumn(
            coluna_destino,
            F.expr(f"try_cast(_{coluna_destino}_t3 as decimal(18,2))") * F.col(f"_{coluna_destino}_mult")
        )
        .withColumn(coluna_destino, F.col(coluna_destino).cast("decimal(18,2)"))
        # 6. zero ou negativo é ausência de dado
        .withColumn(
            coluna_destino,
            F.when(F.col(coluna_destino) <= 0, None).otherwise(F.col(coluna_destino))
        )
        .drop(
            f"_{coluna_destino}_t1", f"_{coluna_destino}_t2",
            f"_{coluna_destino}_mult", f"_{coluna_destino}_t3",
        )
    )
```

```python
from pyspark.sql import functions as F
from pyspark.sql.window import Window

df_mapa = spark.table("catalogo_filmes.silver.tb_mapa_id_filme")

# 1. Dedup de reingestão
window_dedup_ingestao = Window.partitionBy("id").orderBy(F.col("ingestion_datetime").desc())

df_bronze_raw = (
    spark.table("catalogo_filmes.bronze.tb_movies_metrics")
    .withColumn("_rn", F.row_number().over(window_dedup_ingestao))
    .filter(F.col("_rn") == 1)
    .drop("_rn")
)

# 2. Traduz o id original para o id canônico
df_bronze_metrics = (
    df_bronze_raw
    .join(df_mapa, df_bronze_raw.id == df_mapa.id_filme_original, "left")
    .withColumn("id", F.coalesce(F.col("id_filme_canonico"), df_bronze_raw.id))
    .drop("id_filme_canonico", "id_filme_original")
)

df_silver_metricas = (
    df_bronze_metrics
    # FIX 1: se o valor bruto tiver qualquer letra, é texto vazado (sinopse) -> NULL direto
    .withColumn(
        "_popularidade_bruta",
        F.when(F.col("popularity").rlike("[A-Za-z]"), None).otherwise(F.col("popularity"))
    )
    .withColumn(
        "_popularidade_normalizada",
        F.when(
            F.col("_popularidade_bruta").contains(",") & F.col("_popularidade_bruta").contains("."),
            F.regexp_replace(F.col("_popularidade_bruta"), ",", "")
        ).when(
            F.col("_popularidade_bruta").contains(",") & ~F.col("_popularidade_bruta").contains("."),
            F.regexp_replace(F.col("_popularidade_bruta"), ",", ".")
        ).otherwise(F.col("_popularidade_bruta"))
    )
    .withColumn(
        "_popularidade_limpa",
        F.regexp_replace(F.col("_popularidade_normalizada"), r"[^0-9.\-]", "")
    )
    .withColumn("popularidade", F.expr("try_cast(_popularidade_limpa as double)"))

    .withColumn("nota_media_tmdb", F.expr("try_cast(vote_average as double)"))
    .withColumn("qtd_votos_tmdb", F.expr("try_cast(vote_count as int)"))
    .withColumn("nota_media_imdb", F.expr("try_cast(averageRating as double)"))
    .withColumn("qtd_votos_imdb", F.expr("try_cast(numVotes as int)"))

    .withColumn(
        "nota_media_tmdb",
        F.when((F.col("nota_media_tmdb") < 0) | (F.col("nota_media_tmdb") > 10), None)
         .otherwise(F.col("nota_media_tmdb"))
    )
    .withColumn(
        "nota_media_imdb",
        F.when((F.col("nota_media_imdb") < 0) | (F.col("nota_media_imdb") > 10), None)
         .otherwise(F.col("nota_media_imdb"))
    )
    .withColumn("qtd_votos_tmdb", F.when(F.col("qtd_votos_tmdb") < 0, None).otherwise(F.col("qtd_votos_tmdb")))
    .withColumn("qtd_votos_imdb", F.when(F.col("qtd_votos_imdb") < 0, None).otherwise(F.col("qtd_votos_imdb")))
    .withColumn("popularidade", F.when(F.col("popularidade") < 0, None).otherwise(F.col("popularidade")))

    # FIX 2: valor inteiro "redondo" numa faixa plausível de ano = ano_lancamento vazado -> NULL
    .withColumn(
        "popularidade",
        F.when(
            (F.col("popularidade") == F.floor(F.col("popularidade"))) &
            (F.col("popularidade") >= 1870) & (F.col("popularidade") <= 2030),
            None
        ).otherwise(F.col("popularidade"))
    )

    .select(
        F.col("id").alias("id_filme"),
        "popularidade", "nota_media_tmdb", "qtd_votos_tmdb", "nota_media_imdb", "qtd_votos_imdb",
    )
)

# 3. Dedup final por id_filme: ids diferentes que viraram o mesmo canônico geram linhas
#    repetidas. A ordenação prioriza a linha que TEM métrica preenchida.
window_dedup_canonico = Window.partitionBy("id_filme").orderBy(
    F.col("popularidade").isNotNull().desc(),
    F.col("nota_media_tmdb").isNotNull().desc(),
    F.col("nota_media_imdb").isNotNull().desc(),
)

df_silver_metricas = (
    df_silver_metricas
    .withColumn("_rn", F.row_number().over(window_dedup_canonico))
    .filter(F.col("_rn") == 1)
    .drop("_rn")
)

df_silver_metricas.write.format("delta").mode("overwrite").option("overwriteSchema", "true").saveAsTable("catalogo_filmes.silver.tb_metricas_engajamento")

display(df_silver_metricas)
```

```python
from pyspark.sql import functions as F

df_mapa = spark.table("catalogo_filmes.silver.tb_mapa_id_filme")

df_bronze_raw = spark.table("catalogo_filmes.bronze.tb_movies_reviews")

# Traduz o id original para o id canônico ANTES do select/dropDuplicates,
# pra que avaliações dos ids duplicados se consolidem no mesmo filme
df_bronze_reviews = (
    df_bronze_raw
    .join(df_mapa, df_bronze_raw.id == df_mapa.id_filme_original, "left")
    .withColumn("id", F.coalesce(F.col("id_filme_canonico"), df_bronze_raw.id))
    .drop("id_filme_canonico", "id_filme_original")
)

df_silver_avaliacoes = (
    df_bronze_reviews
    .select(
        F.col("id").alias("id_filme"),
        F.col("nome").alias("nome_usuario"),
        F.col("nota").alias("nota_usuario"),
        F.col("comentario").alias("comentario_usuario"),
    )
    # Remove duplicatas EXATAS (mesma combinação filme+usuário+nota+comentário).
    # Também resolve as cópias de reingestão da Bronze, já que linhas reingeridas são idênticas.
    .dropDuplicates(["id_filme", "nome_usuario", "nota_usuario", "comentario_usuario"])

    # Conversão segura + nota fora da escala 0-10 vira NULL (mantém a linha, invalida só o valor)
    .withColumn("nota_usuario", F.expr("try_cast(nota_usuario as double)"))
    .withColumn(
        "nota_usuario",
        F.when((F.col("nota_usuario") < 0) | (F.col("nota_usuario") > 10), None)
         .otherwise(F.col("nota_usuario"))
    )

    # Comentário vazio ou só espaço em branco vira o texto padrão.
    # coalesce(coluna, "") trata também o caso de vir NULL puro, não só string vazia/espaços.
    .withColumn(
        "comentario_usuario",
        F.when(
            F.trim(F.coalesce(F.col("comentario_usuario"), F.lit(""))) == "",
            "Sem comentário"
        ).otherwise(F.col("comentario_usuario"))
    )
)

df_silver_avaliacoes.write.format("delta").mode("overwrite").option("overwriteSchema", "true").saveAsTable("catalogo_filmes.silver.tb_avaliacoes_usuarios")

display(df_silver_avaliacoes)
```

```python
from pyspark.sql import functions as F

df_mapa = spark.table("catalogo_filmes.silver.tb_mapa_id_filme")
df_bronze_raw = spark.table("catalogo_filmes.bronze.tb_credits_and_tags")

df_bronze_credits = (
    df_bronze_raw
    .join(df_mapa, df_bronze_raw.id == df_mapa.id_filme_original, "left")
    .withColumn("id", F.coalesce(F.col("id_filme_canonico"), df_bronze_raw.id))
    .drop("id_filme_canonico", "id_filme_original")
)

# Domínio FECHADO de gêneros válidos (catálogo TMDB + variantes IMDb).
# Tudo que não estiver nessa lista é resíduo de Column Shift / falha de parsing.
generos_validos = [
    "Action", "Adventure", "Animation", "Biography", "Comedy", "Crime",
    "Documentary", "Drama", "Family", "Fantasy", "Film-Noir", "History",
    "Horror", "Music", "Musical", "Mystery", "Romance", "Science Fiction",
    "Sci-Fi", "Sport", "Thriller", "TV Movie", "War", "Western",
]

df_silver_generos = (
    df_bronze_credits
    .select(F.col("id").alias("id_filme"), F.col("genres"))
    # Padroniza separador: ";" vira "," pra dividir sempre pelo mesmo delimitador
    .withColumn("genres_normalizado", F.regexp_replace(F.col("genres"), ";", ","))
    # split + explode: cada gênero vira uma linha (conforme pedido no enunciado)
    .withColumn("genero", F.explode(F.split(F.col("genres_normalizado"), ",")))
    # Remove o lixo de escape do CSV quebrado: aspas e barras invertidas
    .withColumn("genero", F.trim(F.regexp_replace(F.col("genero"), r'[\\"]', "")))
    # WHITELIST: só sobrevive o que é gênero de verdade. Elimina de uma vez
    # sinopses, idiomas, nomes de pessoas, produtoras e números deslocados.
    .filter(F.col("genero").isin(generos_validos))
    .select("id_filme", F.col("genero").alias("nome_genero"))
    .dropDuplicates()
)

df_silver_generos.write.format("delta").mode("overwrite").option("overwriteSchema", "true").saveAsTable("catalogo_filmes.silver.tb_generos")

display(df_silver_generos)
```

```python
from pyspark.sql import functions as F

df_mapa = spark.table("catalogo_filmes.silver.tb_mapa_id_filme")
df_bronze_raw = spark.table("catalogo_filmes.bronze.tb_credits_and_tags")

df_bronze_credits = (
    df_bronze_raw
    .join(df_mapa, df_bronze_raw.id == df_mapa.id_filme_original, "left")
    .withColumn("id", F.coalesce(F.col("id_filme_canonico"), df_bronze_raw.id))
    .drop("id_filme_canonico", "id_filme_original")
)


def extrair_entidade(df, coluna_origem, tipo_entidade):
    """Explode uma coluna multi-valorada (cast/directors/writers/production_companies)
    em uma linha por entidade, já rotulada com seu tipo_entidade."""
    return (
        df
        .select(F.col("id").alias("id_filme"), F.col(coluna_origem).alias("bruto"))
        .withColumn("bruto_normalizado", F.regexp_replace(F.col("bruto"), ";", ","))
        .withColumn("nome_entidade", F.explode(F.split(F.col("bruto_normalizado"), ",")))
        # Remove o lixo de escape do CSV quebrado (aspas e barras) antes de avaliar
        .withColumn("nome_entidade", F.trim(F.regexp_replace(F.col("nome_entidade"), r'[\\"]', "")))

        .filter(F.col("nome_entidade") != "")
        # não é valor numérico deslocado
        .filter(~F.col("nome_entidade").rlike(r"^-?\d+(\.\d+)?$"))
        # tem pelo menos uma letra (descarta pontuação solta, símbolos)
        .filter(F.col("nome_entidade").rlike(r"[A-Za-zÀ-ÿ]"))
        # HEURÍSTICA 1: nome de verdade começa com MAIÚSCULA.
        # Fragmentos de sinopse quase sempre começam com minúscula ("and I...", "he devoted...")
        .filter(F.col("nome_entidade").rlike(r"^[A-ZÀ-Þ0-9]"))
        # HEURÍSTICA 2: nome não é frase — limita tamanho e quantidade de palavras
        .filter(F.length(F.col("nome_entidade")) <= 50)
        .filter(F.size(F.split(F.col("nome_entidade"), r"\s+")) <= 6)
        # HEURÍSTICA 3: nome não tem pontuação de frase
        .filter(~F.col("nome_entidade").rlike(r"[?!]"))

        # Padroniza capitalização (depois dos filtros, que dependem da caixa original)
        .withColumn("nome_entidade", F.initcap(F.col("nome_entidade")))
        .withColumn("tipo_entidade", F.lit(tipo_entidade))
        .select("id_filme", "nome_entidade", "tipo_entidade")
    )


df_atores      = extrair_entidade(df_bronze_credits, "cast", "Ator")
df_diretores   = extrair_entidade(df_bronze_credits, "directors", "Diretor")
df_roteiristas = extrair_entidade(df_bronze_credits, "writers", "Roteirista")
df_produtoras  = extrair_entidade(df_bronze_credits, "production_companies", "Produtora")

# unionByName junta os 4 DataFrames (mesmas colunas) numa única tabela consolidada
df_silver_pessoas_empresas = (
    df_atores
    .unionByName(df_diretores)
    .unionByName(df_roteiristas)
    .unionByName(df_produtoras)
    .dropDuplicates()
)

df_silver_pessoas_empresas.write.format("delta").mode("overwrite").option("overwriteSchema", "true").saveAsTable("catalogo_filmes.silver.tb_pessoas_empresas")

display(df_silver_pessoas_empresas)
```
