```python
def gerar_surrogate_key(coluna):
    """Gera uma BIGINT determinística a partir de uma chave natural, usando SHA-256 (sha2)."""
    hash_hex = F.sha2(coluna.cast("string"), 256)
    return F.conv(F.substring(hash_hex, 1, 15), 16, 10).cast("bigint")


# 1. Lê as 3 tabelas Silver necessárias
df_silver_movies = spark.table("catalogo_filmes.silver.tb_info_filmes")
df_silver_financeiro = spark.table("catalogo_filmes.silver.tb_financeiro_filmes")
df_silver_metricas = spark.table("catalogo_filmes.silver.tb_metricas_engajamento")

# 2. Filtra só os filmes com status "Lançado" — define o grão da fato
df_filmes_lancados = df_silver_movies.filter(F.col("status_filme") == "Lançado").select("id_filme")

# 3. LEFT JOIN com financeiro e métricas — mantém 1 linha por filme mesmo se faltar dado
df_fact_base = (
    df_filmes_lancados
    .join(df_silver_financeiro, on="id_filme", how="left")
    .join(df_silver_metricas, on="id_filme", how="left")
)

# 4. Gera a surrogate key e ajusta os tipos de cada métrica conforme o PDF
df_fact_movies = (
    df_fact_base
    .withColumn("sk_movie_id", gerar_surrogate_key(F.col("id_filme")))
    .select(
        "sk_movie_id",
        F.col("orcamento_usd").cast("decimal(18,2)").alias("orcamento_usd"),
        F.col("receita_usd").cast("decimal(18,2)").alias("receita_usd"),
        F.col("lucro_usd").cast("decimal(18,2)").alias("lucro_usd"),
        F.col("orcamento_brl").cast("decimal(18,2)").alias("orcamento_brl"),
        F.col("receita_brl").cast("decimal(18,2)").alias("receita_brl"),
        F.col("lucro_brl").cast("decimal(18,2)").alias("lucro_brl"),
        F.col("popularidade").cast("double").alias("popularidade"),
        F.col("nota_media_tmdb").cast("double").alias("nota_media_tmdb"),
        F.col("qtd_votos_tmdb").cast("int").alias("qtd_votos_tmdb"),
        F.col("nota_media_imdb").cast("double").alias("nota_media_imdb"),
        F.col("qtd_votos_imdb").cast("int").alias("qtd_votos_imdb"),
    )
)

# 5. Grava na Gold
df_fact_movies.write.format("delta").mode("overwrite").saveAsTable("catalogo_filmes.gold.fact_movies_performance")

display(df_fact_movies)
```

Dimensão de filmes. A surrogate key é gerada por hash SHA-256 da chave natural, e o motivo
dessa escolha em vez de monotonically_increasing_id() ou row_number() está no README, seção
"Camada Gold: modelo dimensional". O dropDuplicates aqui é redundante de propósito e o
raciocínio está na seção "Avisos do otimizador do Databricks".

```python
def gerar_surrogate_key(coluna):
    """Gera uma BIGINT determinística a partir de uma chave natural, usando SHA-256 (sha2)."""
    hash_hex = F.sha2(coluna.cast("string"), 256)
    return F.conv(F.substring(hash_hex, 1, 15), 16, 10).cast("bigint")


df_silver_movies = spark.table("catalogo_filmes.silver.tb_info_filmes")

df_dim_movies = (
    df_silver_movies
    .dropDuplicates(["id_filme"])
    .withColumn("sk_movie_id", gerar_surrogate_key(F.col("id_filme")))
    .select(
        "sk_movie_id",
        F.col("id_filme").cast("string").alias("id_filme"),
        F.col("titulo").cast("string").alias("titulo"),
        F.expr("try_cast(data_lancamento as date)").alias("data_lancamento"),
        F.expr("try_cast(ano_lancamento as int)").alias("ano_lancamento"),
        F.expr("try_cast(duracao_minutos as int)").alias("duracao_minutos"),
        F.col("idioma_original").cast("string").alias("idioma_original"),
        F.col("status_filme").cast("string").alias("status_filme"),
        F.col("sinopse").cast("string").alias("sinopse"),
    )
)

df_dim_movies.write.format("delta").mode("overwrite").saveAsTable("catalogo_filmes.gold.dim_movies")

display(df_dim_movies)
```

Dimensão de pessoas físicas (Ator, Diretor, Roteirista). A surrogate key usa hash composto
de nome e tipo, não só do nome, e o motivo está no README, seção "Camada Gold: modelo
dimensional". O distinct aqui é obrigatório, porque a origem tem grão de uma linha por
participação e não por pessoa, o que também está explicado na seção "Avisos do otimizador
do Databricks".


```python
def gerar_surrogate_key(coluna):
    """Gera uma BIGINT determinística a partir de uma chave natural, usando SHA-256 (sha2)."""
    hash_hex = F.sha2(coluna.cast("string"), 256)
    return F.conv(F.substring(hash_hex, 1, 15), 16, 10).cast("bigint")


df_silver_pessoas = spark.table("catalogo_filmes.silver.tb_pessoas_empresas")

df_dim_people = (
    df_silver_pessoas
    .filter(F.col("tipo_entidade") != "Produtora")
    .select(
        F.col("nome_entidade").cast("string").alias("nome_pessoa"),
        F.col("tipo_entidade").cast("string").alias("tipo_pessoa"),
    )
    .distinct()
    .withColumn(
        "sk_person_id",
        gerar_surrogate_key(F.concat_ws("|", F.col("nome_pessoa"), F.col("tipo_pessoa")))
    )
    .select("sk_person_id", "nome_pessoa", "tipo_pessoa")
)

df_dim_people.write.format("delta").mode("overwrite").saveAsTable("catalogo_filmes.gold.dim_people")

display(df_dim_people)
```

Dimensão de produtoras. Diferente da dim_people, aqui a surrogate key usa só o nome, porque
o tipo é sempre o mesmo depois do filtro e não acrescenta nada à chave. A geração das
surrogate keys está explicada no README, seção "Camada Gold: modelo dimensional".


```python
def gerar_surrogate_key(coluna):
    """Gera uma BIGINT determinística a partir de uma chave natural, usando SHA-256 (sha2)."""
    hash_hex = F.sha2(coluna.cast("string"), 256)
    return F.conv(F.substring(hash_hex, 1, 15), 16, 10).cast("bigint")


df_silver_pessoas = spark.table("catalogo_filmes.silver.tb_pessoas_empresas")

df_dim_companies = (
    df_silver_pessoas
    .filter(F.col("tipo_entidade") == "Produtora")
    .select(F.col("nome_entidade").cast("string").alias("nome_produtora"))
    .distinct()
    .withColumn("sk_company_id", gerar_surrogate_key(F.col("nome_produtora")))
    .select("sk_company_id", "nome_produtora")
)

df_dim_companies.write.format("delta").mode("overwrite").saveAsTable("catalogo_filmes.gold.dim_companies")

display(df_dim_companies)
```

Resume as avaliações individuais numa métrica por filme. O sk_movie_id vem de join real com
a dim_movies, e não do hash direto, porque aqui ele é chave estrangeira e precisa apontar
para uma linha que existe de fato. A média usa avg(), que ignora nulos sozinho, então as
notas invalidadas na Silver não puxam o resultado para baixo. O tratamento das notas fora
de escala está no README, seção "Tratamento por tabela", em tb_avaliacoes_usuarios.

```python
def gerar_surrogate_key(coluna):
    """Gera uma BIGINT determinística a partir de uma chave natural, usando SHA-256 (sha2)."""
    hash_hex = F.sha2(coluna.cast("string"), 256)
    return F.conv(F.substring(hash_hex, 1, 15), 16, 10).cast("bigint")


df_silver_avaliacoes = spark.table("catalogo_filmes.silver.tb_avaliacoes_usuarios")
df_dim_movies = spark.table("catalogo_filmes.gold.dim_movies")

df_dim_reviews = (
    df_silver_avaliacoes
    .groupBy("id_filme")
    .agg(
        F.count(F.lit(1)).alias("qtd_avaliacoes_usuarios"),
        F.round(F.avg("nota_usuario"), 2).alias("nota_media_usuarios"),
    )
    .join(df_dim_movies.select("id_filme", "sk_movie_id"), on="id_filme", how="inner")
    .withColumn("sk_review_id", gerar_surrogate_key(F.col("id_filme")))
    .select(
        "sk_review_id",
        "sk_movie_id",
        F.col("qtd_avaliacoes_usuarios").cast("int").alias("qtd_avaliacoes_usuarios"),
        F.col("nota_media_usuarios").cast("double").alias("nota_media_usuarios"),
    )
)

df_dim_reviews.write.format("delta").mode("overwrite").saveAsTable("catalogo_filmes.gold.dim_reviews")

display(df_dim_reviews)
```

Catálogo único de gêneros. A chave natural aqui é o próprio nome, porque gênero não tem
identificador de origem, ele veio de uma coluna de texto explodida na Silver. A geração da
surrogate key está no README, seção "Camada Gold: modelo dimensional", e a limpeza por
lista de válidos que garante a qualidade dessa lista está em "Domínio fechado contra
domínio aberto".

```python
def gerar_surrogate_key(coluna):
    hash_hex = F.sha2(coluna.cast("string"), 256)
    return F.conv(F.substring(hash_hex, 1, 15), 16, 10).cast("bigint")

df_silver_generos = spark.table("catalogo_filmes.silver.tb_generos")

df_dim_genres = (
    df_silver_generos
    .select(F.col("nome_genero").cast("string").alias("nome_genero"))
    .distinct()
    .withColumn("sk_genre_id", gerar_surrogate_key(F.col("nome_genero")))
    .select("sk_genre_id", "nome_genero")
)

df_dim_genres.write.format("delta").mode("overwrite").saveAsTable("catalogo_filmes.gold.dim_genres")

display(df_dim_genres)
```

Tabela ponte entre filmes e gêneros. Ela existe porque a relação é N para N e ligar os
gêneros direto na fato duplicaria as métricas financeiras a cada gênero do filme. Só guarda
o par de chaves estrangeiras, nenhum dado descritivo. Ao contrário das dimensões, aqui as
surrogate keys vêm de join real com dim_movies e dim_genres, porque são chaves estrangeiras
e precisam apontar para linhas existentes. O modelo dimensional está no README, seção
"Camada Gold: modelo dimensional".

```python
from pyspark.sql import functions as F

df_silver_generos = spark.table("catalogo_filmes.silver.tb_generos")
df_dim_movies = spark.table("catalogo_filmes.gold.dim_movies")
df_dim_genres = spark.table("catalogo_filmes.gold.dim_genres")

df_bridge_movie_genre = (
    df_silver_generos
    # Junta pela chave natural (id_filme / nome_genero) pra "pegar emprestado" as surrogate keys
    .join(df_dim_movies.select("id_filme", "sk_movie_id"), on="id_filme", how="inner")
    .join(df_dim_genres.select("nome_genero", "sk_genre_id"), on="nome_genero", how="inner")
    .select("sk_movie_id", "sk_genre_id")
    .dropDuplicates()  # evita linha repetida se o mesmo gênero aparecer 2x pro mesmo filme na origem
)

df_bridge_movie_genre.write.format("delta").mode("overwrite").saveAsTable("catalogo_filmes.gold.bridge_movie_genre")

display(df_bridge_movie_genre)
```

Tabela ponte entre filmes e pessoas físicas, que resolve a relação N para N sem duplicar o grão da fato. A diferença em relação à ponte de gêneros está no segundo join: aqui ele é composto por nome e tipo juntos, porque a dim_people usa a combinação dos dois como chave. Se o join fosse só pelo nome, uma pessoa que atua como Ator num filme e Diretor em outro traria as duas linhas da dimensão e duplicaria a ponte. O modelo dimensional e o motivo da chave composta estão no README, seção "Camada Gold: modelo dimensional".

```python
df_silver_pessoas = spark.table("catalogo_filmes.silver.tb_pessoas_empresas")
df_dim_movies = spark.table("catalogo_filmes.gold.dim_movies")
df_dim_people = spark.table("catalogo_filmes.gold.dim_people")
df_dim_companies = spark.table("catalogo_filmes.gold.dim_companies")

df_bridge_movie_person = (
    df_silver_pessoas
    .filter(F.col("tipo_entidade") != "Produtora")  # só pessoa física aqui
    .withColumnRenamed("nome_entidade", "nome_pessoa")
    .withColumnRenamed("tipo_entidade", "tipo_pessoa")
    .join(df_dim_movies.select("id_filme", "sk_movie_id"), on="id_filme", how="inner")
    # join composto: precisa bater NOME e TIPO juntos, já que dim_people usa a combinação dos dois como chave
    .join(df_dim_people.select("nome_pessoa", "tipo_pessoa", "sk_person_id"), on=["nome_pessoa", "tipo_pessoa"], how="inner")
    .select("sk_movie_id", "sk_person_id")
    .dropDuplicates()
)

df_bridge_movie_person.write.format("delta").mode("overwrite").saveAsTable("catalogo_filmes.gold.bridge_movie_person")

display(df_bridge_movie_person)
```

Tabela ponte entre filmes e produtoras, mesma função das outras duas: resolver a relação N para N sem multiplicar as métricas da fato. Aqui o join com a dimensão é simples, só pelo nome, porque a dim_companies não tem chave composta, diferente da dim_people. Os dois inner join garantem que nenhuma chave estrangeira aponte para uma linha inexistente. O modelo dimensional está no README, seção "Camada Gold: modelo dimensional"

```python
df_bridge_movie_company = (
    df_silver_pessoas
    .filter(F.col("tipo_entidade") == "Produtora")  # só pessoa jurídica aqui
    .withColumnRenamed("nome_entidade", "nome_produtora")
    .join(df_dim_movies.select("id_filme", "sk_movie_id"), on="id_filme", how="inner")
    .join(df_dim_companies.select("nome_produtora", "sk_company_id"), on="nome_produtora", how="inner")
    .select("sk_movie_id", "sk_company_id")
    .dropDuplicates()
)

df_bridge_movie_company.write.format("delta").mode("overwrite").saveAsTable("catalogo_filmes.gold.bridge_movie_company")

display(df_bridge_movie_company)
```

Carrega as tabelas da Gold que alimentam as consultas de negócio. Essa célula não transforma nada, só abre as referências para as perguntas seguintes, evitando repetir os spark.table em cada uma delas.

```python
from pyspark.sql import functions as F
from pyspark.sql.window import Window

df_dim_movies = spark.table("catalogo_filmes.gold.dim_movies")
df_fact = spark.table("catalogo_filmes.gold.fact_movies_performance")
df_dim_genres = spark.table("catalogo_filmes.gold.dim_genres")
df_bridge_genre = spark.table("catalogo_filmes.gold.bridge_movie_genre")
df_dim_people = spark.table("catalogo_filmes.gold.dim_people")
df_bridge_person = spark.table("catalogo_filmes.gold.bridge_movie_person")
df_dim_companies = spark.table("catalogo_filmes.gold.dim_companies")
df_bridge_company = spark.table("catalogo_filmes.gold.bridge_movie_company")
```

Responde questão 1

```python
display(
    df_fact.agg(F.sum("receita_brl").alias("receita_total_brl"))
)
```

Responde questão 2

```python
display(
    df_dim_movies
    .join(df_fact, on="sk_movie_id")
    .select("titulo", "popularidade")
    .orderBy(F.col("popularidade").desc())
    .limit(5)
)
```

Responde qestão 3

```python
display(
    df_bridge_genre
    .join(df_dim_genres, on="sk_genre_id")
    .groupBy("nome_genero")
    .agg(F.count("sk_movie_id").alias("qtd_filmes"))
    .orderBy(F.col("qtd_filmes").desc())
)
```

Responde questão 4

```python
window_receita = Window.orderBy(F.col("receita_usd").desc())

display(
    df_dim_movies
    .join(df_fact, on="sk_movie_id")
    .select("titulo", "receita_usd", "receita_brl")
    .withColumn("posicao_ranking", F.rank().over(window_receita))
    .orderBy("posicao_ranking")
    .limit(10)
)
```

Calcula a data para usar no ator com mais filmes nos últimos 2 anos

```python
# checa da data de lancamento para depois ver se eh maior que 2 anos ou nao
data_mais_recente = (
    df_dim_movies
    .filter((F.col("status_filme") == "Lançado") & (F.col("data_lancamento") <= F.current_date()))
    .agg(F.max("data_lancamento"))
    .first()[0]
)

print("Data de referência:", data_mais_recente)
```

Responde questão 5

```python
df_filmes_2anos = df_dim_movies.filter(
    (F.col("data_lancamento") <= F.lit(data_mais_recente)) &
    (F.col("data_lancamento") >= F.add_months(F.lit(data_mais_recente), -24))
)

display(
    df_filmes_2anos
    .join(df_bridge_person, on="sk_movie_id")
    .join(df_dim_people.filter(F.col("tipo_pessoa") == "Ator"), on="sk_person_id")
    .groupBy("nome_pessoa")
    .agg(
        F.countDistinct("sk_movie_id").alias("qtd_participacoes"),
        F.concat_ws(", ", F.collect_set("titulo")).alias("lista_filmes")
    )
    .orderBy(F.col("qtd_participacoes").desc())
    .limit(1)
)
```

Responde questão 6

```python
df_filmes_5anos = df_dim_movies.filter(
    (F.col("data_lancamento") <= F.lit(data_mais_recente)) &
    (F.col("data_lancamento") >= F.add_months(F.lit(data_mais_recente), -60))
)

display(
    df_filmes_5anos
    .join(df_fact, on="sk_movie_id")
    .join(df_bridge_company, on="sk_movie_id")
    .join(df_dim_companies, on="sk_company_id")
    .groupBy("nome_produtora")
    .agg(F.sum("lucro_usd").alias("lucro_total_usd"))
    .orderBy(F.col("lucro_total_usd").desc())
    .limit(1)
)
```
