```python
def gerar_surrogate_key(coluna):
    """Gera uma BIGINT determinística a partir de uma chave natural, usando SHA-256 (sha2)."""
    hash_hex = F.sha2(coluna.cast("string"), 256)
    return F.conv(F.substring(hash_hex, 1, 15), 16, 10).cast("bigint")


# 1. Lê as 3 tabelas Silver necessárias
df_silver_movies = spark.table("catalogo_filmes.silver.tb_info_filmes")
df_silver_financeiro = spark.table("catalogo_filmes.silver.tb_financeiro_filmes")
df_silver_metricas = spark.table("catalogo_filmes.silver.tb_metricas_engajamento")

# 2. Filtra só os filmes com status "Lançado" — define o grão da fato
df_filmes_lancados = df_silver_movies.filter(F.col("status_filme") == "Lançado").select("id_filme")

# 3. LEFT JOIN com financeiro e métricas — mantém 1 linha por filme mesmo se faltar dado
df_fact_base = (
    df_filmes_lancados
    .join(df_silver_financeiro, on="id_filme", how="left")
    .join(df_silver_metricas, on="id_filme", how="left")
)

# 4. Gera a surrogate key e ajusta os tipos de cada métrica conforme o PDF
df_fact_movies = (
    df_fact_base
    .withColumn("sk_movie_id", gerar_surrogate_key(F.col("id_filme")))
    .select(
        "sk_movie_id",
        F.col("orcamento_usd").cast("decimal(18,2)").alias("orcamento_usd"),
        F.col("receita_usd").cast("decimal(18,2)").alias("receita_usd"),
        F.col("lucro_usd").cast("decimal(18,2)").alias("lucro_usd"),
        F.col("orcamento_brl").cast("decimal(18,2)").alias("orcamento_brl"),
        F.col("receita_brl").cast("decimal(18,2)").alias("receita_brl"),
        F.col("lucro_brl").cast("decimal(18,2)").alias("lucro_brl"),
        F.col("popularidade").cast("double").alias("popularidade"),
        F.col("nota_media_tmdb").cast("double").alias("nota_media_tmdb"),
        F.col("qtd_votos_tmdb").cast("int").alias("qtd_votos_tmdb"),
        F.col("nota_media_imdb").cast("double").alias("nota_media_imdb"),
        F.col("qtd_votos_imdb").cast("int").alias("qtd_votos_imdb"),
    )
)

# 5. Grava na Gold
df_fact_movies.write.format("delta").mode("overwrite").saveAsTable("catalogo_filmes.gold.fact_movies_performance")

display(df_fact_movies)
```

```python
def gerar_surrogate_key(coluna):
    """Gera uma BIGINT determinística a partir de uma chave natural, usando SHA-256 (sha2)."""
    hash_hex = F.sha2(coluna.cast("string"), 256)
    return F.conv(F.substring(hash_hex, 1, 15), 16, 10).cast("bigint")


df_silver_pessoas = spark.table("catalogo_filmes.silver.tb_pessoas_empresas")

df_dim_people = (
    df_silver_pessoas
    # dim_people é só PESSOA FÍSICA — exclui "Produtora" (isso vai pra dim_companies)
    .filter(F.col("tipo_entidade") != "Produtora")
    .select(
        F.col("nome_entidade").cast("string").alias("nome_pessoa"),
        F.col("tipo_entidade").cast("string").alias("tipo_pessoa"),
    )
    .distinct()
    # hash de 2 colunas juntas: uma mesma pessoa que é Ator num filme e Diretor em outro
    # gera surrogate keys DIFERENTES para cada combinação nome+tipo (são "papéis" distintos)
    .withColumn(
        "sk_person_id",
        gerar_surrogate_key(F.concat_ws("|", F.col("nome_pessoa"), F.col("tipo_pessoa")))
    )
    .select("sk_person_id", "nome_pessoa", "tipo_pessoa")
)

df_dim_people.write.format("delta").mode("overwrite").saveAsTable("catalogo_filmes.gold.dim_people")

display(df_dim_people)
```

```python
def gerar_surrogate_key(coluna):
    """Gera uma BIGINT determinística a partir de uma chave natural, usando SHA-256 (sha2)."""
    hash_hex = F.sha2(coluna.cast("string"), 256)
    return F.conv(F.substring(hash_hex, 1, 15), 16, 10).cast("bigint")


df_silver_pessoas = spark.table("catalogo_filmes.silver.tb_pessoas_empresas")

df_dim_companies = (
    df_silver_pessoas
    # dim_companies é só PESSOA JURÍDICA — só as linhas marcadas como "Produtora"
    .filter(F.col("tipo_entidade") == "Produtora")
    .select(F.col("nome_entidade").cast("string").alias("nome_produtora"))
    .distinct()  # catálogo único, conforme pedido
    .withColumn("sk_company_id", gerar_surrogate_key(F.col("nome_produtora")))
    .select("sk_company_id", "nome_produtora")
)

df_dim_companies.write.format("delta").mode("overwrite").saveAsTable("catalogo_filmes.gold.dim_companies")

display(df_dim_companies)
```

```python
def gerar_surrogate_key(coluna):
    """Gera uma BIGINT determinística a partir de uma chave natural, usando SHA-256 (sha2)."""
    hash_hex = F.sha2(coluna.cast("string"), 256)
    return F.conv(F.substring(hash_hex, 1, 15), 16, 10).cast("bigint")


df_silver_avaliacoes = spark.table("catalogo_filmes.silver.tb_avaliacoes_usuarios")
df_dim_movies = spark.table("catalogo_filmes.gold.dim_movies")

df_dim_reviews = (
    df_silver_avaliacoes
    # 1. Resume as avaliações individuais numa métrica por filme
    .groupBy("id_filme")
    .agg(
        F.count(F.lit(1)).alias("qtd_avaliacoes_usuarios"),
        F.round(F.avg("nota_usuario"), 2).alias("nota_media_usuarios"),  # avg() já ignora NULL sozinho
    )
    # 2. Junta com dim_movies só para pegar o sk_movie_id (usa a MESMA fórmula de hash,
    #    então bate certinho com a chave já gerada na dim_movies)
    .join(df_dim_movies.select("id_filme", "sk_movie_id"), on="id_filme", how="inner")
    # 3. Gera a PK própria dessa dimensão
    .withColumn("sk_review_id", gerar_surrogate_key(F.col("id_filme")))
    .select(
        "sk_review_id",
        "sk_movie_id",
        F.col("qtd_avaliacoes_usuarios").cast("int").alias("qtd_avaliacoes_usuarios"),
        F.col("nota_media_usuarios").cast("double").alias("nota_media_usuarios"),
    )
)

df_dim_reviews.write.format("delta").mode("overwrite").saveAsTable("catalogo_filmes.gold.dim_reviews")

display(df_dim_reviews)
```

```python
from pyspark.sql import functions as F

df_silver_generos = spark.table("catalogo_filmes.silver.tb_generos")
df_dim_movies = spark.table("catalogo_filmes.gold.dim_movies")
df_dim_genres = spark.table("catalogo_filmes.gold.dim_genres")

df_bridge_movie_genre = (
    df_silver_generos
    # Junta pela chave natural (id_filme / nome_genero) pra "pegar emprestado" as surrogate keys
    .join(df_dim_movies.select("id_filme", "sk_movie_id"), on="id_filme", how="inner")
    .join(df_dim_genres.select("nome_genero", "sk_genre_id"), on="nome_genero", how="inner")
    .select("sk_movie_id", "sk_genre_id")
    .dropDuplicates()  # evita linha repetida se o mesmo gênero aparecer 2x pro mesmo filme na origem
)

df_bridge_movie_genre.write.format("delta").mode("overwrite").saveAsTable("catalogo_filmes.gold.bridge_movie_genre")

display(df_bridge_movie_genre)
```

```python
df_silver_pessoas = spark.table("catalogo_filmes.silver.tb_pessoas_empresas")
df_dim_movies = spark.table("catalogo_filmes.gold.dim_movies")
df_dim_people = spark.table("catalogo_filmes.gold.dim_people")
df_dim_companies = spark.table("catalogo_filmes.gold.dim_companies")

df_bridge_movie_person = (
    df_silver_pessoas
    .filter(F.col("tipo_entidade") != "Produtora")  # só pessoa física aqui
    .withColumnRenamed("nome_entidade", "nome_pessoa")
    .withColumnRenamed("tipo_entidade", "tipo_pessoa")
    .join(df_dim_movies.select("id_filme", "sk_movie_id"), on="id_filme", how="inner")
    # join composto: precisa bater NOME e TIPO juntos, já que dim_people usa a combinação dos dois como chave
    .join(df_dim_people.select("nome_pessoa", "tipo_pessoa", "sk_person_id"), on=["nome_pessoa", "tipo_pessoa"], how="inner")
    .select("sk_movie_id", "sk_person_id")
    .dropDuplicates()
)

df_bridge_movie_person.write.format("delta").mode("overwrite").saveAsTable("catalogo_filmes.gold.bridge_movie_person")

display(df_bridge_movie_person)
```

```python
df_bridge_movie_company = (
    df_silver_pessoas
    .filter(F.col("tipo_entidade") == "Produtora")  # só pessoa jurídica aqui
    .withColumnRenamed("nome_entidade", "nome_produtora")
    .join(df_dim_movies.select("id_filme", "sk_movie_id"), on="id_filme", how="inner")
    .join(df_dim_companies.select("nome_produtora", "sk_company_id"), on="nome_produtora", how="inner")
    .select("sk_movie_id", "sk_company_id")
    .dropDuplicates()
)

df_bridge_movie_company.write.format("delta").mode("overwrite").saveAsTable("catalogo_filmes.gold.bridge_movie_company")

display(df_bridge_movie_company)
```

```python
# checa da data de lancamento para depois ver se eh maior que 2 anos ou nao
data_mais_recente = (
    df_dim_movies
    .filter((F.col("status_filme") == "Lançado") & (F.col("data_lancamento") <= F.current_date()))
    .agg(F.max("data_lancamento"))
    .first()[0]
)

print("Data de referência:", data_mais_recente)
```

```python
from pyspark.sql import functions as F

df_filmes_2anos = df_dim_movies.filter(
    (F.col("data_lancamento") <= F.lit(data_mais_recente)) &
    (F.col("data_lancamento") >= F.add_months(F.lit(data_mais_recente), -24))
)

display(
    df_filmes_2anos
    .join(df_bridge_person, on="sk_movie_id")
    .join(df_dim_people.filter(F.col("tipo_pessoa") == "Ator"), on="sk_person_id")
    .groupBy("nome_pessoa")
    .agg(
        F.countDistinct("sk_movie_id").alias("qtd_participacoes"),
        F.concat_ws(", ", F.collect_set("titulo")).alias("lista_filmes")
    )
    .orderBy(F.col("qtd_participacoes").desc())
    .limit(1)
)
```

```python
df_filmes_5anos = df_dim_movies.filter(
    (F.col("data_lancamento") <= F.lit(data_mais_recente)) &
    (F.col("data_lancamento") >= F.add_months(F.lit(data_mais_recente), -60))
)

display(
    df_filmes_5anos
    .join(df_fact, on="sk_movie_id")
    .join(df_bridge_company, on="sk_movie_id")
    .join(df_dim_companies, on="sk_company_id")
    .groupBy("nome_produtora")
    .agg(F.sum("lucro_usd").alias("lucro_total_usd"))
    .orderBy(F.col("lucro_total_usd").desc())
    .limit(1)
)
```
